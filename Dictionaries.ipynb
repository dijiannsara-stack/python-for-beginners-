```python
Course = {
    "Java": "4 weeks only",
    "Python": "6 weeks only",
    "DoNet": "7 weeks only",
    "C++": "8 weeks only",
    "Php": "6 weeks only",
    "Android": "10 weeks only"
}
```

```python
print(Course)
print("Enter the learning period for Java:", Course ["Java"])
print("Enter the learning period for Python:", Course ["Python"])
print("Enter the learning period for DoNet", Course ["DoNet"])
print("Enter the learning period for C++", Course ["C++"])
print("Enter the learning period for Php", Course ["Php"])
print("Enter the learning period for 'Android':", Course ["Android"])
```
