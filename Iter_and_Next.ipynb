```python
def course_generate():
  yield ("Java")
  yield ("Python")
  yield ("DoNet")
  yield ("C++")
  yield ("Php")
  yield ("Vb.Net")
  yield ("Android")


```

```python
course = course_generate()
print(next(course))
print(next(course))
print(next(course))
print(next(course))
print(next(course))
print(next(course))
print(next(course))
```
